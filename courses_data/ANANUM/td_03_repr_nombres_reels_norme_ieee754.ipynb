## Exercice 03.05 (facultatif)

Le discriminant des deux trinômes suivants est nul :


$$x^{2}+0.2x+0.01$$

$$x^{2}+1.4x+0.49$$

Par conséquent le nombre de solution retourné par le programme ci-dessous devrait être égale à 1. Testez le programme.

Proposez une solution.

```python
def nbsolutions(a,b,c):
    Delta = b**2 - 4*a*c
    if Delta == 0:
        nb = 1
    elif Delta > 0:
        nb = 2
    else:
        nb = 0
    return nb
```
